### Reading the Dataset and Separating Inputs from the Target


The first three columns—City, Age, and Salary—form the input matrix `X`. The fourth column, Purchased, forms the target vector `Y`. Missing numerical values appear as `NaN`.


```python
#----------------------------------Reading the dataset-----------------------
dataset = pd.read_csv('salary.csv')
# pd is the alias of the Pandas library imported.
X = dataset.iloc[:, :-1].values
Y = dataset.iloc[:,3].values
print(X)
print(Y)
```

## 3. Handling Missing Numerical Data


`SimpleImputer` calculates the mean of the selected Age and Salary columns and replaces their missing values. The printed example fits the imputer on the complete dataset to demonstrate the operation. In a predictive workflow, fit imputation using training data only to avoid data leakage.


```python
# handling of missing data
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values = np.nan, strategy = 'mean')
imputer = imputer.fit(X[:,1:3])
X[:,1:3] = imputer.transform(X[:,1:3])
print(X)
```

## 4. Label Encoding the City Attribute


`LabelEncoder` converts the City names into integer labels. These integers identify categories; they do not imply that one city is greater than another.


```python
#label encoding categorical attribute city
from sklearn.preprocessing import LabelEncoder
labelEncoder_X = LabelEncoder()
X[:, 0] = labelEncoder_X.fit_transform(X[:, 0])
print(X[:, 0])
```

## 5. One-Hot Encoding the City Attribute


`OneHotEncoder` converts the integer-coded City attribute into separate binary columns. `ColumnTransformer` applies this transformation to column 0 and passes the remaining columns through unchanged.


```python
#One-hot encoding categorical attribute city
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
print(X)
#Alternate way to do One-hot encoding is mentioned in the end of this file
```

## 6. Label Encoding the Purchased Attribute


The Purchased target contains `Yes` and `No`. Label encoding converts these two classes into numerical values so they can be used by machine-learning algorithms.


```python
#label encoding purchased attribute
labelEncoder_Y = LabelEncoder()
Y = labelEncoder_Y.fit_transform(Y)
print(Y)
```

## 7. Splitting the Dataset into Training and Test Sets


The dataset is divided into 80% training data and 20% test data. `random_state=0` preserves the split demonstrated in the printed chapter and makes the result reproducible.


```python
#Splitting of Dataset into training and testing
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 0)
print(X_train)
print(X_test)
print(Y_train)
print(Y_test)
```

## 8. Normalizing the Features


L1 normalization rescales each observation so that the sum of the absolute values in that row equals 1. This is row-wise normalization, not the same operation as mapping each feature to a 0–1 range.


```python
#Feature Scaling
from sklearn import preprocessing
X_train = preprocessing.normalize(X_train, norm='l1')
X_test = preprocessing.normalize(X_test, norm='l1')
print(X_train)
print(X_test)
```

## 9. Standardizing the Features


The original split is recreated because the earlier cell changed `X_train` and `X_test` through normalization. `StandardScaler` learns the training means and standard deviations with `fit_transform()` and applies those same values to the test set with `transform()`.


```python
#Standardizing the data
#lets do split again as we have already applied normalization on X_train and X_test
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 0)
#apply standarization
from sklearn.preprocessing import StandardScaler
scale_X = StandardScaler()
X_train = scale_X.fit_transform(X_train)
X_test = scale_X.transform(X_test)
print(X_train)
print(X_test)
```

## 10. Alternative One-Hot Encoding with Pandas


`pd.get_dummies()` provides a concise alternative for converting the City column into binary indicator columns. With `drop_first=True`, one reference category is omitted to avoid redundant dummy variables.


```python
#Alternate way to do One-hot encoding
#Importing of desired libraries
import numpy as np
import pandas as pd
#---------------------------Reading the dataset----------------------
dataset = pd.read_csv('salary.csv')
# pd is the alias of the Pandas library imported.
dataset = pd.get_dummies(dataset, columns=["City"], drop_first=True)
print(dataset)
```

## Key Takeaways

- Missing numerical values can be replaced using `SimpleImputer`.
- Categorical attributes must be converted into a numerical representation.
- Label encoding is appropriate for a binary target, while one-hot encoding avoids imposing an artificial order on nominal input categories.
- Training and test data serve different purposes and should remain separate during model development.
- Normalization and standardization are different feature-scaling operations.
- Preprocessing parameters should be learned from training data and then applied unchanged to test data.
