## 1. Importing the Libraries


```python
# Importing the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Loading the Dataset

Download `StipendData.csv` from this GitHub folder. Run the following cell and select the downloaded file when prompted.

> The upload widget works in Google Colab. The uploaded file remains available only during the current runtime session.


```python
from google.colab import files

# Upload StipendData.csv from your local system.
uploaded = files.upload()
```

```python
# Loading the dataset
dataset = pd.read_csv("StipendData.csv")

print("Dataset shape:", dataset.shape)
display(dataset.head())
```

Before building a model, verify that the expected columns were loaded and inspect the first few observations.


## 3. Creating the Feature Matrix and Target Vector


```python
# Creating the feature matrix and target vector
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values

print("Feature matrix shape:", X.shape)
print("Target vector shape:", y.shape)
print("\nFeature values:\n", X)
print("\nTarget values:\n", y)
```

`X` contains the independent variable, years of research experience. `y` contains the dependent variable, stipend. Scikit-learn expects `X` to be two-dimensional, even when the model uses only one feature.


## 4. Splitting the Data into Training and Test Sets


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=1/3,
    random_state=0,
)

print("Training observations:", len(X_train))
print("Test observations:", len(X_test))
print("\nX_train:\n", X_train)
print("\nX_test:\n", X_test)
print("\ny_train:\n", y_train)
print("\ny_test:\n", y_test)
```

The training set is used to learn the regression line. The test set is held back to evaluate predictions on unseen observations. `random_state=0` preserves the split used in the printed chapter and makes the result reproducible.


## 5. Creating the Linear Regression Model


```python
from sklearn.linear_model import LinearRegression

# Creating the LinearRegression object
linear_regressor = LinearRegression()
```

## 6. Training the Model


```python
# Fitting the model to the training data
linear_regressor.fit(X_train, y_train)

intercept = linear_regressor.intercept_
slope = linear_regressor.coef_[0]

print("Intercept:", intercept)
print("Coefficient:", slope)
print(f"Fitted equation: stipend = {intercept:.3f} + {slope:.3f} × experience")
```

The fitted model follows:

$$\hat{y}=\theta_0+\theta_1x$$

- `intercept_` represents the predicted stipend when research experience is zero.
- `coef_[0]` represents the predicted change in stipend for one additional year of research experience.

The coefficient describes an association in this dataset; it does not by itself establish causation.


## 7. Generating Predictions


```python
# Predicting the test-set results
y_pred = linear_regressor.predict(X_test)

prediction_results = pd.DataFrame({
    "Research Experience": X_test.ravel(),
    "Actual Stipend": y_test,
    "Predicted Stipend": y_pred,
    "Residual": y_test - y_pred,
})

display(prediction_results.sort_values("Research Experience"))
```

A residual is the difference between an actual and predicted value. Positive residuals lie above the fitted line; negative residuals lie below it.


## 8. Visualizing the Training Results


```python
# Sorting x-values prevents the plotted regression line from connecting points out of order.
train_order = np.argsort(X_train.ravel())

plt.figure(figsize=(7, 4.5))
plt.scatter(
    X_train,
    y_train,
    color="red",
    edgecolor="black",
    label="Training observations",
)
plt.plot(
    X_train.ravel()[train_order],
    linear_regressor.predict(X_train)[train_order],
    color="blue",
    linewidth=2,
    label="Regression line",
)
plt.title("Simple Linear Regression: Training Data")
plt.xlabel("Years of Research Experience")
plt.ylabel("Stipend")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

## 9. Visualizing the Test Results


```python
# Creating a smooth x-range for the fitted line
x_line = np.linspace(X.min(), X.max(), 200).reshape(-1, 1)
y_line = linear_regressor.predict(x_line)

plt.figure(figsize=(7, 4.5))
plt.scatter(
    X_test,
    y_test,
    color="red",
    edgecolor="black",
    label="Test observations",
)
plt.plot(
    x_line,
    y_line,
    color="blue",
    linewidth=2,
    label="Regression line",
)
plt.title("Simple Linear Regression: Test Data")
plt.xlabel("Years of Research Experience")
plt.ylabel("Stipend")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

## 10. Evaluating Model Performance


```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

evaluation = pd.DataFrame({
    "Metric": ["MAE", "MSE", "RMSE", "R²"],
    "Value": [mae, mse, rmse, r2],
})

display(evaluation)
```

- **MAE** reports the average absolute prediction error.
- **MSE** gives greater weight to larger errors.
- **RMSE** expresses the typical error in the original target units.
- **$R^2$** measures the proportion of variation explained by the model relative to a mean-based baseline.

Because this dataset is small, these values should be interpreted as an instructional demonstration rather than conclusive evidence of real-world performance.


## Key Takeaways

- Simple linear regression models the relationship between one input feature and one continuous target.
- Scikit-learn's `LinearRegression` learns the intercept and coefficient from the training data.
- Test data should remain separate until model evaluation.
- MAE, MSE, RMSE, and $R^2$ provide complementary information.
- A fitted relationship represents association, not necessarily causation.
- Predictions beyond the observed experience range should be treated cautiously.
